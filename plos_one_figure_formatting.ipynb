```python
from PIL import Image
import os

def convert_jpg_to_plos_tiff(input_path, output_path=None, dpi=(300, 300)):
    # Open the JPG image
    img = Image.open(input_path)

    # Convert to RGB if not already
    if img.mode != 'RGB':
        img = img.convert('RGB')

    # Set output path if not specified
    if output_path is None:
        base, _ = os.path.splitext(input_path)
        output_path = f"{base}.tif"

    # Save as TIFF with LZW compression and 300 DPI
    img.save(output_path, format='TIFF', dpi=dpi, compression='tiff_lzw')

    print(f"Saved PLOS-compliant TIFF to: {output_path}")
```

```python
def inspect_image(path):
    with Image.open(path) as img:
        print(f"Mode: {img.mode}")
        print(f"Size: {img.size} pixels")
        print(f"Info: {img.info}")
```

```python
convert_jpg_to_plos_tiff('drafted_figures/fig1_aramai_point_rocks.jpg', output_path= 'final_figures/fig1.tiff')
```

```python
inspect_image('final_figures/fig1.tiff')
```

```python
import matplotlib.pyplot as plt

def show_tiff(path):
    img = Image.open(path)
    plt.imshow(img)
    plt.axis('off')  # Hide axes
    plt.show()
```

```python
show_tiff('final_figures/fig1.tiff')
```
